```python
crosstab_res=pd.crosstab(df['category'], df['isBestSeller'])
```

```python
crosstab_res
```

```python
sorted_crosstab_res = crosstab_res.sort_values(by=True, ascending=False)
```

```python
sorted_crosstab_res
```

```python
from scipy.stats import chi2_contingency
chi2_statistic, chi2_p_value, _, _= chi2_contingency(crosstab_res)
chi2_p_value<0.05
```

```python
def outlify(i):
    Q1 = i.quantile(0.25)
    Q3 = i.quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = i[(i < lower_bound) | (i > upper_bound)]
    
    return outliers

price_outliers = outlify(df["price"])

df = df[~df.index.isin(price_outliers.index)]
```

```python
sns.violinplot(data=df[df['category'].isin(df['category'].value_counts().head(20).index)], x='category', y='price', palette='coolwarm')
```

```python
sns.barplot(data=df[df['category'].isin(df['category'].value_counts().head(10).index)], x='category', y='price', palette='coolwarm')
```

```python
df_numerical = df.select_dtypes('number')
correlation_matrix = df_numerical.corr()

plt.figure(figsize=(18,15))
mask = np.triu(correlation_matrix)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', mask=mask)
plt.show()
```
