## **1. Objective**

The objective of this analysis is to explore relationships between key product variables in the Amazon UK marketplace, with a focus on best-seller status, pricing, categories, brands, and customer ratings.

From a business perspective, this study aims to identify patterns that can support category strategy, pricing decisions, and product positioning.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency, pearsonr, probplot
from scipy.stats.contingency import association

pd.set_option("display.max_columns", None)
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

current_dir = os.getcwd()

while not os.path.exists(os.path.join(current_dir, "data", "raw")):
    parent = os.path.dirname(current_dir)
    if parent == current_dir:
        raise FileNotFoundError("Project root with data/raw not found")
    current_dir = parent

os.chdir(current_dir)

df = pd.read_csv("data/raw/amz_uk_price_prediction_dataset.csv")
df.info()
df.head()
```

## **2. Part 1: Analyzing Best-Seller Trends Across Product Categories**

### **2.1.1. Crosstab**

```python
crosstab_cat_bestseller = pd.crosstab(df["category"], df["isBestSeller"])
crosstab_cat_bestseller.head()
```

**Insight**

The distribution of best-sellers across categories shows that only a very small proportion of products achieve best-seller status, regardless of category.

This suggests that becoming a best-seller is a highly competitive outcome, likely driven by factors beyond category alone, such as visibility, reviews, and demand dynamics.

### **2.1.2. Proportion of best-sellers by category**

```python
best_seller_prop = (
    pd.crosstab(df["category"], df["isBestSeller"], normalize="index") * 100
)

best_seller_prop = best_seller_prop.rename(columns={False: "not_best_seller_pct", True: "best_seller_pct"})
best_seller_prop = best_seller_prop.sort_values("best_seller_pct", ascending=False)

best_seller_prop.head(10)
```

**Insight**

Some categories, such as Grocery, Smart Home, and Health & Personal Care, show relatively higher proportions of best-sellers (above ~5%), compared to most categories.

This indicates that certain categories have structural advantages for best-seller emergence, possibly due to:
- High purchase frequency
- Standardized products
- Strong recurring demand

In contrast, other categories show significantly lower best-seller proportions, suggesting more fragmented competition.

### **2.1.3. Chi-square test**

```python
chi2, p, dof, expected = chi2_contingency(crosstab_cat_bestseller)
print("Chi-square statistic:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
```

**Insight**

The Chi-square test yields a p-value of 0.0, indicating that the relationship between product category and best-seller status is statistically significant.

This means that best-seller distribution is not random across categories and that category plays a role in influencing product success.

### **2.1.4. Cramér’s V**

```python
cramers_v = association(crosstab_cat_bestseller, method="cramer")
print("Cramér's V:", cramers_v)
```

**Insight**

Cramér’s V = 0.12 indicates a weak association between category and best-seller status.

From a business perspective:
- Category matters, but it is not the dominant driver of best-seller success

### **2.1.5. Stacked bar chart**

```python
top20_categories = df["category"].value_counts().head(20).index
df_top20 = df[df["category"].isin(top20_categories)]

stacked_data = pd.crosstab(df_top20["category"], df_top20["isBestSeller"], normalize="index")
stacked_data.plot(kind="bar", stacked=True, figsize=(12, 6))
plt.title("Best-Seller Share Across Top 20 Categories")
plt.ylabel("Proportion")
plt.xlabel("Category")
plt.xticks(rotation=75)
plt.legend(title="isBestSeller")
plt.show()
```

**Insight**

The stacked bar chart confirms that the proportion of best-sellers remains consistently low across all major categories.

Even in high-performing categories, best-sellers represent only a small fraction of listings, reinforcing the idea that:
- The marketplace is highly competitive
- Success is concentrated in a small subset of products

This highlights the importance of differentiation and positioning strategies.

## **3. Part 2: Exploring Product Prices and Ratings Across Categories and Brands**

### **3.2.1. Remove outliers with IQR**

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_no_outliers = df[(df["price"] >= lower_bound) & (df["price"] <= upper_bound)].copy()

print("Original rows:", len(df))
print("Rows without outliers:", len(df_no_outliers))
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

**Insight**

Outlier removal reduced the dataset from ~2.44M to ~2.11M observations, eliminating extreme price values above ~100.

This step is critical because:
- It allows for a more realistic view of typical product pricing
- It prevents extreme values from distorting averages and distributions

From a business perspective, this reflects the core market rather than niche or luxury extremes.

### **3.2.2. Top 20 categories by count**

```python
top20_categories = df_no_outliers["category"].value_counts().head(20).index
df_top20_no_outliers = df_no_outliers[df_no_outliers["category"].isin(top20_categories)]

plt.figure(figsize=(14, 6))
sns.violinplot(data=df_top20_no_outliers, x="category", y="price")
plt.xticks(rotation=75)
plt.title("Price Distribution Across Top 20 Categories (Without Outliers)")
plt.xlabel("Category")
plt.ylabel("Price")
plt.show()
```

**Insight**

Price distributions vary significantly across categories, with clear differences in:
- Median prices
- Spread and variability
- Presence of multiple price tiers

Some categories show tight distributions (standardized pricing), while others display wide variability, indicating:
- Different pricing strategies
- Different levels of product differentiation

This suggests that pricing strategy should be category-specific rather than universal.

### **3.2.3. Highest Median Price Category**

```python
category_median_price = (
    df_no_outliers.groupby("category")["price"]
    .median()
    .sort_values(ascending=False)
)

category_median_price.head(10)
```

**Insight**

Technology-related categories (e.g., Desktop PCs, Graphics Cards, Tablets) dominate the highest median prices.

This reflects:
- Higher production costs
- Strong perceived value
- Lower price sensitivity among buyers

These categories operate in a premium pricing environment, where margins may be higher but competition is also more specialized.

### **3.2.4. Bar Chart Of Average Price For Top 10 Categories By Count**

```python
top10_categories = df_no_outliers["category"].value_counts().head(10).index
avg_price_top10 = (
    df_no_outliers[df_no_outliers["category"].isin(top10_categories)]
    .groupby("category")["price"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))
sns.barplot(x=avg_price_top10.values, y=avg_price_top10.index)
plt.title("Average Price Across Top 10 Categories (Without Outliers)")
plt.xlabel("Average Price")
plt.ylabel("Category")
plt.show()
```

**Insight**

Among high-volume categories, Sports & Outdoors and Birthday Gifts show relatively higher average prices compared to others.

This suggests that:
- High-volume categories can still sustain moderate pricing levels
- Demand scale does not necessarily imply low pricing

This creates opportunities for balancing volume and margin strategies.

### **3.2.5. Highest Average Price**

```python
category_avg_price = (
    df_no_outliers.groupby("category")["price"]
    .mean()
    .sort_values(ascending=False)
)

category_avg_price.head(10)
```

**Insight**

Among high-volume categories, Sports & Outdoors and Birthday Gifts show relatively higher average prices compared to others.

This suggests that:
- High-volume categories can still sustain moderate pricing levels
- Demand scale does not necessarily imply low pricing

This creates opportunities for balancing volume and margin strategies.

### **3.2.6. Top 10 Categories By Count**

```python
top10_categories = df_no_outliers["category"].value_counts().head(10).index
df_top10_no_outliers = df_no_outliers[df_no_outliers["category"].isin(top10_categories)]

plt.figure(figsize=(14, 6))
sns.boxplot(data=df_top10_no_outliers, x="category", y="stars")
plt.xticks(rotation=75)
plt.title("Rating Distribution Across Top 10 Categories")
plt.xlabel("Category")
plt.ylabel("Stars")
plt.show()
```

**Insight**

Most categories show high median ratings (around 4–4.5 stars), indicating generally positive customer satisfaction across the platform.

However, the presence of many 0-star values suggests:
- Products without ratings (not necessarily poorly rated)
- Potential bias in the dataset

This must be interpreted carefully to avoid misleading conclusions.

### **3.2.7. Highest Median Rating**

```python
category_median_rating = (
    df_no_outliers.groupby("category")["stars"]
    .median()
    .sort_values(ascending=False)
)

category_median_rating.head(10)
```

**Insight**

Categories such as Computer Memory, Food & Drink, Toys, and Office Products achieve the highest median ratings (~4.6–4.7).

These categories likely benefit from:
- Clear expectations
- Standardized product quality
- Lower complexity in user experience

This suggests that customer satisfaction is easier to achieve in categories with well-defined product standards.

## **4. Part 3: Investigating the Interplay Between Product Prices and Ratings**

### **4.3.1. Correlation Coefficient**

```python
corr_price_stars = df_no_outliers["price"].corr(df_no_outliers["stars"])
print("Correlation between price and stars:", corr_price_stars)

pearson_corr, pearson_p = pearsonr(df_no_outliers["price"], df_no_outliers["stars"])
print("Pearson correlation:", pearson_corr)
print("p-value:", pearson_p)
```

**Insight**

The correlation between price and rating is weak and slightly negative (-0.08).

This indicates that:
- Higher-priced products do not necessarily receive better ratings
- Price is not a strong predictor of customer satisfaction

From a business perspective:
Premium pricing must be justified by perceived value, not assumed to generate better reviews.

### **4.3.2. Scatter Plot**

```python
sample_df = df_no_outliers.sample(10000, random_state=42)

plt.figure(figsize=(8, 6))
sns.scatterplot(data=sample_df, x="stars", y="price", alpha=0.3)
plt.title("Price vs Rating (Sample, Without Outliers)")
plt.xlabel("Stars")
plt.ylabel("Price")
plt.show()
```

**Insight**

The scatter plot shows a dense cluster of products across all price levels for ratings between 3.5 and 5 stars.

There is no clear upward or downward trend, confirming:
- Weak relationship between price and rating
- High variability within each rating level

Additionally, a large vertical cluster at 0 stars indicates many unrated products, which impacts overall analysis.

### **4.3.3. Correlation heatmap**

```python
numeric_cols = df_no_outliers.select_dtypes(include=[np.number]).columns
corr_matrix = df_no_outliers[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

**Insight**

The heatmap confirms that correlations between variables are generally weak.

Key observations:
- Price has very low correlation with all variables
- Reviews and ratings show a slight positive relationship
- Purchase activity (boughtInLastMonth) has weak associations

Conclusion:
Marketplace dynamics are complex and cannot be explained by simple linear relationships.

### **4.3.4. QQ Plot For Price**

```python
plt.figure(figsize=(8, 6))
probplot(df_no_outliers["price"], dist="norm", plot=plt)
plt.title("QQ Plot of Price (Without Outliers)")
plt.show()
```

**Insight**

The QQ plot shows a clear deviation from normality, even after removing outliers.

Price distribution remains:
- Right-skewed
- Non-linear
- Heavy-tailed

This indicates that:
Pricing follows a marketplace-driven distribution rather than a normal statistical pattern

Implication:
- Mean is not a reliable metric alone
- Median and percentiles are more meaningful for business decisions

### **4.3.5. Same Analysis With Outliers**

```python
corr_with_outliers = df["price"].corr(df["stars"])
corr_without_outliers = df_no_outliers["price"].corr(df_no_outliers["stars"])

print("Correlation with outliers:", corr_with_outliers)
print("Correlation without outliers:", corr_without_outliers)
```
