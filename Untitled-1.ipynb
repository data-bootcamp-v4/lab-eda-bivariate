```python
# STEP 0: Imports & Load
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, pearsonr, probplot

plt.rcParams["figure.figsize"] = (8, 5)
pd.set_option("display.float_format", lambda v: f"{v:,.3f}")

# Load your dataset (EDIT the path/filename if different)
df = pd.read_csv("amazon_uk_products.csv")

print("Columns:", df.columns.tolist())
df.head()
```

```python
# STEP 1: Clean key columns

# Clean price -> float
df["price"] = (
    df["price"].astype(str)
    .str.replace("£", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.extract(r"([0-9]+(?:\.[0-9]+)?)", expand=False)
    .astype(float)
)

# Clean stars -> float
df["stars"] = (
    df["stars"].astype(str)
    .str.extract(r"([0-9]+(?:\.[0-9]+)?)", expand=False)
    .astype(float)
)

# Normalize isBestSeller -> boolean True/False
truthy = {"true","1","yes","y","t"}
falsy  = {"false","0","no","n","f"}

df["isBestSeller"] = (
    df["isBestSeller"].map(
        lambda x: str(x).strip().lower()
    ).map(lambda s: True if s in truthy else (False if s in falsy else np.nan))
)

# Drop rows with missing essentials for the parts that need them
df_part1 = df.dropna(subset=["category", "isBestSeller"])
df_price = df.dropna(subset=["category", "price"])
df_rate  = df.dropna(subset=["category", "stars"])
```

```python
# STEP 1: Clean key columns

# Clean price -> float
df["price"] = (
    df["price"].astype(str)
    .str.replace("£", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.extract(r"([0-9]+(?:\.[0-9]+)?)", expand=False)
    .astype(float)
)

# Clean stars -> float
df["stars"] = (
    df["stars"].astype(str)
    .str.extract(r"([0-9]+(?:\.[0-9]+)?)", expand=False)
    .astype(float)
)

# Normalize isBestSeller -> boolean True/False
truthy = {"true","1","yes","y","t"}
falsy  = {"false","0","no","n","f"}

df["isBestSeller"] = (
    df["isBestSeller"].map(
        lambda x: str(x).strip().lower()
    ).map(lambda s: True if s in truthy else (False if s in falsy else np.nan))
)

# Drop rows with missing essentials for the parts that need them
df_part1 = df.dropna(subset=["category", "isBestSeller"])
df_price = df.dropna(subset=["category", "price"])
df_rate  = df.dropna(subset=["category", "stars"])
```

```python
# STEP 1A: Crosstab & proportion of best-sellers per category
crosstab = pd.crosstab(df_part1["category"], df_part1["isBestSeller"])
crosstab.head()
```

```python
# Proportion of best-sellers (True) per category
if True in crosstab.columns:
    pos_col = True
else:
    # fallback if your data ended up 1/0 for any reason
    pos_col = 1 if 1 in crosstab.columns else crosstab.columns[-1]

prop_bestseller = (crosstab[pos_col] / crosstab.sum(axis=1)).fillna(0)

# Optional: filter out tiny categories to avoid noise
MIN_COUNT_FOR_PROP = 20
valid = crosstab.sum(axis=1) >= MIN_COUNT_FOR_PROP
prop_bestseller_sorted = prop_bestseller[valid].sort_values(ascending=False)
prop_bestseller_sorted.head(10)
```

```python
# STEP 1B: Chi-square test + Cramér's V
chi2, p, dof, expected = chi2_contingency(crosstab)
n = crosstab.values.sum()
r, k = crosstab.shape
phi2 = chi2 / n
cramers_v = np.sqrt(phi2 / max(1, min(k-1, r-1)))

print(f"Chi-square = {chi2:,.3f}, p-value = {p:.4g}, dof = {dof}")
print(f"Cramér's V = {cramers_v:.3f}")
```

```python
# STEP 1C: Stacked bar of bestseller proportion for top categories by count
top15_by_count = crosstab.sum(axis=1).sort_values(ascending=False).head(15).index
crosstab_top = crosstab.loc[top15_by_count]
crosstab_top_norm = crosstab_top.div(crosstab_top.sum(axis=1), axis=0).fillna(0)

ax = crosstab_top_norm.plot(kind="bar", stacked=True, figsize=(12,6))
ax.set_title("Best-Seller Distribution by Category (Top 15 by Count)")
ax.set_ylabel("Proportion")
ax.set_xlabel("Category")
ax.legend(title="isBestSeller")
plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()
```

```python
# STEP 2A: Outlier removal on price (IQR)
p = df_price["price"].dropna()
Q1, Q3 = p.quantile([0.25, 0.75])
IQR = Q3 - Q1
low, high = Q1 - 1.5*IQR, Q3 + 1.5*IQR

df_no_out = df_price[(df_price["price"] >= low) & (df_price["price"] <= high)].copy()
print(f"Kept {len(df_no_out):,} rows after IQR filtering (from {len(df_price):,}).")
```

```python
# STEP 2B: Violin plot (Top 20 categories by count)
top20 = df_no_out["category"].value_counts().head(20).index
plt.figure(figsize=(14,6))
sns.violinplot(x="category", y="price", data=df_no_out[df_no_out["category"].isin(top20)], cut=0)
plt.title("Price Distribution by Category (Top 20 by Count, Outliers Removed)")
plt.xlabel("Category")
plt.ylabel("Price")
plt.xticks(rotation=70, ha="right")
plt.tight_layout()
plt.show()
```

```python
# STEP 2C: Highest median price category (no top-k filter)
median_prices = df_no_out.groupby("category")["price"].median().sort_values(ascending=False)
print("Highest median price category:")
median_prices.head(5)
```

```python
# STEP 2D: Average price for top 10 categories by count
top10_by_count = df_no_out["category"].value_counts().head(10).index
avg_price_top10 = df_no_out[df_no_out["category"].isin(top10_by_count)] \
                    .groupby("category")["price"].mean().sort_values(ascending=False)

ax = avg_price_top10.plot(kind="bar", figsize=(10,5))
ax.set_title("Average Price by Category (Top 10 by Count, Outliers Removed)")
ax.set_ylabel("Average Price")
ax.set_xlabel("Category")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

print("Category with highest average price (within top-10-by-count):")
avg_price_top10.head(1)
```

```python
# STEP 2E: Highest average price across ALL categories
avg_price_all = df_no_out.groupby("category")["price"].mean().sort_values(ascending=False)
print("Highest average price category (no filter):")
avg_price_all.head(5)
```

```python
# STEP 2F: Ratings distribution by category (Top 10 by count)
# Build a ratings-clean df aligned with df_no_out
df_rate_aligned = df_no_out.dropna(subset=["stars"]).copy()

top10_rate = df_rate_aligned["category"].value_counts().head(10).index
plt.figure(figsize=(12,6))
sns.boxplot(x="category", y="stars", data=df_rate_aligned[df_rate_aligned["category"].isin(top10_rate)])
plt.title("Ratings by Category (Top 10 by Count)")
plt.xlabel("Category")
plt.ylabel("Rating (Stars)")
plt.xticks(rotation=70, ha="right")
plt.tight_layout()
plt.show()
```

```python
# STEP 2G: Highest median rating (no filter)
median_ratings_all = df_rate.groupby("category")["stars"].median().sort_values(ascending=False)
print("Highest median rating category (no filter):")
median_ratings_all.head(5)
```

```python
# STEP 3D: QQ plot for price normality
plt.figure(figsize=(6,6))
probplot(df_no_out["price"].dropna(), dist="norm", plot=plt)
plt.title("QQ Plot — Price (Outliers Removed)")
plt.tight_layout()
plt.show()

# (Optional) Log-price QQ plot (often closer to normal if prices are right-skewed)
plt.figure(figsize=(6,6))
probplot(np.log1p(df_no_out["price"].dropna()), dist="norm", plot=plt)
plt.title("QQ Plot — log(1+Price)")
plt.tight_layout()
plt.show()
```
